```python
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get(
    "https://guides.library.harvard.edu/hks/think_tank_search/US"
)
```

```python
soup = BeautifulSoup(response.content, parser='html.parser')
list_items = soup.find_all('li', attrs={'class': None})
think_tanks = []
for list_item in list_items:
    a_tag = list_item.find('a', attrs={'aria-label': None})
    if a_tag is None or "javascript" in str(a_tag):
        continue
    think_tanks.append([a_tag.get('href'), list_item.text])
len(think_tanks)
```

```python
import re
from pypdf import PdfReader
```

```python
reader = PdfReader("../ignore/content.pdf")
```

```python
top_us_think_tank_pages = range(84,88)
top_us_think_tanks = []
for top_us_think_tank_page in top_us_think_tank_pages:
    page = reader.pages[top_us_think_tank_page].extract_text()
    top_us_think_tanks += [line.strip() for line in page.split('\n') if re.match(r"\d{1,3}\..+", line)]

for top_us_think_tank in top_us_think_tanks:
    print(top_us_think_tank)
```

```python
top_us_think_tank_pages = [58] + list(range(67,74)) + list(range(84,88)) + list(range(112,190)) + list(range(200,255))
#  + list(range(190,193)) # Best New Think Tanks
#  + list(range(193,200)) # Best Policy Study-Report Produced by a Think Tank
#  + list(range(255,256)) # Best Regional Studies Center (Free Standing)
#  + list(range(256,258)) # Best Regional Studies Center (University-Affiliated)
flag = re.compile(r"\(United States\)")
think_tanks = set()
for top_us_think_tank_page in top_us_think_tank_pages:
    # print(top_us_think_tank_page)
    page_lines = reader.pages[top_us_think_tank_page].extract_text().split('\n')
    # print("page:\n", page)
    think_tanks |= {re.sub(r"\d{0,3}\.?", "", line).strip() for line in page_lines if flag.search(line)}
    

for think_tank in think_tanks:
    print(think_tank)
```
